```python
# Address Class
class Address:
    def __init__(self, floor_no, street_name,
                 city, state, country):
        self.floor_no = floor_no
        self.street_name = street_name
        self.city = city
        self.state = state
        self.country = country

    def display_address(self):
        return (f"{self.floor_no}, {self.street_name}, "
                f"{self.city}, {self.state}, {self.country}")


# Parent Class
class Employee:
    company_name = "ABC Company"

    def __init__(self, emp_id, emp_name,
                 salary, address):
        self.emp_id = emp_id
        self.emp_name = emp_name
        self.salary = salary
        self.address = address      # HAS-A Relationship

    @staticmethod
    def company_info():
        return "ABC Company Training Portal"

    def display(self):
        print(f"Employee ID : {self.emp_id}")
        print(f"Employee Name : {self.emp_name}")
        print(f"Salary : {self.salary}")
        print(f"Address : {self.address.display_address()}")


# Child Class 1
class JuniorEngineer(Employee):
    def __init__(self, emp_id, emp_name,
                 salary, address,
                 assessment_score, feedback):
        super().__init__(emp_id, emp_name,
                         salary, address)
        self.assessment_score = assessment_score
        self.feedback = feedback.title()

    # Method Overriding
    def display(self):
        super().display()
        print("Role : Junior Engineer")
        print(f"Assessment Score : {self.assessment_score}")
        print(f"Feedback : {self.feedback}")


# Child Class 2
class SoftwareEngineer(Employee):
    def __init__(self, emp_id, emp_name,
                 salary, address,
                 project_name):
        super().__init__(emp_id, emp_name,
                         salary, address)
        self.project_name = project_name

    # Method Overriding
    def display(self):
        super().display()
        print("Role : Software Engineer")
        print(f"Project Name : {self.project_name}")


# Child Class 3
class Trainer(Employee):
    def __init__(self, emp_id, emp_name,
                 salary, address,
                 skills, certifications):
        super().__init__(emp_id, emp_name,
                         salary, address)
        self.skills = skills
        self.certifications = certifications

    # Method Overriding
    def display(self):
        super().display()
        print("Role : Trainer")
        print("Skills :", ", ".join(self.skills))
        print("Certifications :", ", ".join(self.certifications))


# Dictionary Usage
class FeedbackManager:
    feedback_records = {}

    @staticmethod
    def add_feedback(emp_name, feedback):
        FeedbackManager.feedback_records[emp_name] = feedback.title()

    @staticmethod
    def show_feedbacks():
        print("\nFeedback Records")
        for emp, fb in FeedbackManager.feedback_records.items():
            print(f"{emp} : {fb}")


# Polymorphism
def show_employee_details(employee):
    employee.display()
    print("-" * 40)


# Driver Code
address1 = Address(
    3,
    "MG Road",
    "Hyderabad",
    "Telangana",
    "India"
)

je = JuniorEngineer(
    "E101",
    "Ramya",
    30000,
    address1,
    90,
    "excellent performance"
)

se = SoftwareEngineer(
    "E102",
    "Rahul",
    60000,
    address1,
    "Training Management System"
)

trainer = Trainer(
    "T101",
    "Suresh",
    80000,
    address1,
    ["Python", "Django", "SQL"],
    ["AWS", "Azure"]
)

show_employee_details(je)
show_employee_details(se)
show_employee_details(trainer)

FeedbackManager.add_feedback(
    "Ramya",
    "great training session"
)

FeedbackManager.show_feedbacks()
```
