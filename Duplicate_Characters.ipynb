```python
class Task6:

    @staticmethod
    def find_duplicate_characters(text):

        # Dictionary to store character counts
        char_count = {}

        # String Manipulation
        text = text.lower()

        for char in text:
            if char != " ":  # Ignore spaces
                char_count[char] = char_count.get(char, 0) + 1

        print("Duplicate Characters and Their Counts:\n")

        for char, count in char_count.items():
            if count > 1:
                print(f"{char} -> {count}")


# Driver Code
company_name = "ABC Company pvt lmt"

Task6.find_duplicate_characters(company_name)
```
