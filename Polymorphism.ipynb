```python
# Address Class (Composition)
class Address:
    def __init__(self, floor_no, street, city, state, country):
        self.floor_no = floor_no
        self.street = street
        self.city = city
        self.state = state
        self.country = country

    def full_address(self):
        return f"{self.floor_no}, {self.street}, {self.city}, {self.state}, {self.country}"


# Parent Class
class Employee:
    def __init__(self, emp_id, emp_name, salary, address):
        self.emp_id = emp_id
        self.emp_name = emp_name
        self.salary = salary
        self.address = address

    def print_details(self):
        print("Employee Details")
        print(f"ID      : {self.emp_id}")
        print(f"Name    : {self.emp_name}")
        print(f"Salary  : {self.salary}")


# Child Class
class SoftwareEngineer(Employee):
    def __init__(self, emp_id, emp_name, salary,
                 address, project_name):
        super().__init__(emp_id, emp_name, salary, address)
        self.project_name = project_name

    # Method Overriding
    def print_details(self):
        print("Software Engineer Details")
        print(f"ID      : {self.emp_id}")
        print(f"Name    : {self.emp_name}")
        print(f"Project : {self.project_name}")
        print(f"Address : {self.address.full_address()}")


# Child Class
class Trainer(Employee):
    def __init__(self, emp_id, emp_name, salary,
                 address, skills, certifications):
        super().__init__(emp_id, emp_name, salary, address)
        self.skills = skills
        self.certifications = certifications

    # Method Overriding
    def print_details(self):
        print("Trainer Details")
        print(f"ID      : {self.emp_id}")
        print(f"Name    : {self.emp_name}")
        print("Skills :", ", ".join(self.skills))
        print("Certificates :", ", ".join(self.certifications))


# Polymorphism
def show_employee(employee):
    employee.print_details()
    print("-" * 40)


# Driver Code
addr = Address(
    3,
    "MG Road",
    "Hyderabad",
    "Telangana",
    "India"
)

se = SoftwareEngineer(
    "E101",
    "Ramya",
    60000,
    addr,
    "Training Management System"
)

trainer = Trainer(
    "T101",
    "Suresh",
    80000,
    addr,
    ["Python", "Django", "SQL"],
    ["AWS", "Azure"]
)

show_employee(se)
show_employee(trainer)
```
