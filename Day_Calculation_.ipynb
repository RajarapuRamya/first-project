```python
class Task5:
    # Static variable
    DAYS = {
        "Monday": 0,
        "Tuesday": 1,
        "Wednesday": 2,
        "Thursday": 3,
        "Friday": 4,
        "Saturday": 5,
        "Sunday": 6
    }

    # Static method
    @staticmethod
    def find_day(current_day, k):

        # String Manipulation
        current_day = current_day.strip().title()

        if current_day not in Task5.DAYS:
            return "Invalid Day"

        current_index = Task5.DAYS[current_day]

        # Day calculation
        new_index = (current_index + k) % 7

        # Dictionary lookup to get day name
        for day, index in Task5.DAYS.items():
            if index == new_index:
                return day


# Driver Code
day = input("Enter Current Day: ")
k = int(input("Enter K Value: "))

result = Task5.find_day(day, k)

print("Day after", k, "days is:", result)
```
