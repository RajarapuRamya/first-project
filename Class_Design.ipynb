```python
class Employee:
    company_name = "ABC Company"   # Static Variable

    def __init__(self, emp_id, emp_name, salary):
        self.emp_id = emp_id
        self.emp_name = emp_name
        self.salary = salary

    def display(self):
        print(f"Employee ID : {self.emp_id}")
        print(f"Employee Name : {self.emp_name}")
        print(f"Salary : {self.salary}")

    @staticmethod
    def company_info():
        return "ABC Company Training Portal"


class Course:
    def __init__(self, course_id, course_name, duration):
        self.course_id = course_id
        self.course_name = course_name
        self.duration = duration

    def display(self):
        print(f"Course ID : {self.course_id}")
        print(f"Course Name : {self.course_name}")
        print(f"Duration : {self.duration}")


# Inheritance
class TrainingRegistration(Employee):
    def __init__(self, emp_id, emp_name, salary, course):
        super().__init__(emp_id, emp_name, salary)
        self.course = course      # Composition (HAS-A)

    # Method Overriding
    def display(self):
        print("\n--- Registration Details ---")
        print(f"Employee : {self.emp_name}")
        print(f"Course : {self.course.course_name}")
        print(f"Duration : {self.course.duration}")


class Feedback:
    def __init__(self):
        self.feedback_data = {}   # Dictionary Usage

    def add_feedback(self, employee_name, feedback):
        # String Manipulation
        feedback = feedback.strip().title()
        self.feedback_data[employee_name] = feedback

    def show_feedback(self):
        print("\n--- Feedback Records ---")
        for emp, fb in self.feedback_data.items():
            print(f"{emp} : {fb}")


# Polymorphism Example
def show_details(obj):
    obj.display()


# Driver Code
course1 = Course("C101", "Python Programming", "30 Days")

emp1 = TrainingRegistration(
    "E101",
    "Ramya",
    50000,
    course1
)

feedback = Feedback()
feedback.add_feedback("Ramya", "excellent training session")

print(Employee.company_info())
print()

show_details(course1)   # Polymorphism
print()

show_details(emp1)      # Polymorphism
feedback.show_feedback()
```
