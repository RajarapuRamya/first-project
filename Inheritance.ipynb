```python
# Base Class
class Employee:
    company_name = "ABC Company"

    def __init__(self, emp_id, emp_name, salary):
        self.emp_id = emp_id
        self.emp_name = emp_name
        self.salary = salary

    def display(self):
        print(f"\nEmployee ID : {self.emp_id}")
        print(f"Employee Name : {self.emp_name}")
        print(f"Salary : {self.salary}")

    @staticmethod
    def company_info():
        return "ABC Company Training Management System"


# Course Class (Composition)
class Course:
    def __init__(self, course_id, course_name, duration):
        self.course_id = course_id
        self.course_name = course_name
        self.duration = duration

    def display(self):
        print(f"Course : {self.course_name}")
        print(f"Duration : {self.duration}")


# Child Class 1
class JuniorEngineer(Employee):
    def __init__(self, emp_id, emp_name, salary,
                 assessment_score, feedback, course):
        super().__init__(emp_id, emp_name, salary)
        self.assessment_score = assessment_score
        self.feedback = feedback.strip().title()
        self.course = course      # HAS-A relationship

    # Method Overriding
    def display(self):
        super().display()
        print("Role : Junior Engineer")
        print(f"Assessment Score : {self.assessment_score}")
        print(f"Feedback : {self.feedback}")
        print(f"Registered Course : {self.course.course_name}")


# Child Class 2
class SoftwareEngineer(Employee):
    def __init__(self, emp_id, emp_name, salary,
                 project_name, course):
        super().__init__(emp_id, emp_name, salary)
        self.project_name = project_name
        self.course = course

    # Method Overriding
    def display(self):
        super().display()
        print("Role : Software Engineer")
        print(f"Project Name : {self.project_name}")
        print(f"Registered Course : {self.course.course_name}")


# Child Class 3
class Trainer(Employee):
    def __init__(self, emp_id, emp_name, salary,
                 skills, certifications):
        super().__init__(emp_id, emp_name, salary)
        self.skills = skills
        self.certifications = certifications

    # Method Overriding
    def display(self):
        super().display()
        print("Role : Trainer")
        print("Skills :", ", ".join(self.skills))
        print("Certifications :", ", ".join(self.certifications))


# Feedback Management (Dictionary Usage)
class FeedbackManager:
    def __init__(self):
        self.feedback_records = {}

    def add_feedback(self, employee_name, feedback):
        self.feedback_records[employee_name] = feedback.title()

    def show_feedbacks(self):
        print("\n----- Feedback Records -----")
        for name, feedback in self.feedback_records.items():
            print(f"{name} : {feedback}")


# Polymorphism
def employee_details(emp):
    emp.display()


# Driver Code
python_course = Course(
    "C101",
    "Python Programming",
    "30 Days"
)

je = JuniorEngineer(
    "E101",
    "Ramya",
    30000,
    88,
    "good learning experience",
    python_course
)

se = SoftwareEngineer(
    "E102",
    "Rahul",
    60000,
    "Employee Management System",
    python_course
)

trainer = Trainer(
    "T101",
    "Suresh",
    80000,
    ["Python", "Django", "SQL"],
    ["AWS", "Microsoft Certified"]
)

feedback_mgr = FeedbackManager()
feedback_mgr.add_feedback(
    "Ramya",
    "excellent training session"
)

print(Employee.company_info())

employee_details(je)
employee_details(se)
employee_details(trainer)

feedback_mgr.show_feedbacks()
```
